# **ML Models**

```python
# ============================== #
#   TabPFN + ML Ensembles Runner
#   (robust ensemble saving, optional repeated CV, mean±std tables)
# ============================== #
from __future__ import annotations
import os, warnings, json
from pathlib import Path
from shutil import rmtree

import numpy as np
import pandas as pd
import torch
from scipy.special import expit

from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score, roc_auc_score,
    balanced_accuracy_score, matthews_corrcoef, cohen_kappa_score,
    average_precision_score, brier_score_loss, log_loss, confusion_matrix
)
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression, RidgeClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.ensemble import (
    RandomForestClassifier, GradientBoostingClassifier, AdaBoostClassifier,
    BaggingClassifier, ExtraTreesClassifier
)

from tabpfn import TabPFNClassifier
warnings.filterwarnings("ignore")


# ---------- utilities ----------
def _get_effective_device(requested: str = "auto") -> str:
    """
    Returns the device string to use for TabPFN based on user request and availability.
    - 'auto': prefer CUDA if available; TabPFN v2.x doesn't run on MPS, so fall back to CPU.
    - 'cpu' : force CPU.
    - 'cuda': use CUDA if available else CPU.
    - 'mps' : force CPU (TabPFN v2.x not MPS-ready).
    """
    if requested.lower() == "auto":
        return "cuda" if torch.cuda.is_available() else "cpu"
    if requested.lower() == "cuda":
        return "cuda" if torch.cuda.is_available() else "cpu"
    if requested.lower() == "mps":
        return "cpu"
    return "cpu"


def _maybe_import_autotabpfn(use_extensions: bool):
    """Try to import AutoTabPFNClassifier from tabpfn-extensions; return class or None."""
    if not use_extensions:
        print("ℹ️ tabpfn-extensions disabled by user.")
        return None
    try:
        from tabpfn_extensions.post_hoc_ensembles import AutoTabPFNClassifier as _AutoTabPFNClassifier
        print("🔌 AutoTabPFNClassifier detected (tabpfn-extensions).")
        return _AutoTabPFNClassifier
    except Exception as e:
        print(f"ℹ️ tabpfn-extensions AutoTabPFN unavailable ({e}). Proceeding without it.")
        return None


def _sanitize_autogluon_fit_kwargs(kwargs: dict) -> dict:
    """Avoid duplicate args in AutoGluon fit."""
    for dup_key in ("time_limit", "presets"):
        if dup_key in kwargs:
            kwargs.pop(dup_key, None)
            print(f"⚠️  Removed duplicate '{dup_key}' from phe_fit_args to avoid AutoGluon conflict.")
    return kwargs


def _get_probabilities(model, X):
    """
    Return positive-class probabilities with safe fallbacks:
      1) predict_proba[:,1]
      2) decision_function -> sigmoid
      3) predict (0/1)
    """
    if hasattr(model, "predict_proba"):
        proba = model.predict_proba(X)
        if proba.ndim == 2 and proba.shape[1] >= 2:
            return proba[:, 1]
        return proba.ravel()
    elif hasattr(model, "decision_function"):
        scores = model.decision_function(X)
        return expit(scores)
    else:
        preds = model.predict(X)
        return preds.astype(float)


def _compute_all_metrics(y_true: np.ndarray, proba_pos: np.ndarray, threshold: float = 0.5) -> dict:
    eps = 1e-12
    proba_pos = np.clip(proba_pos, eps, 1 - eps)
    y_pred = (proba_pos >= threshold).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred, labels=[0, 1]).ravel()

    sensitivity = recall_score(y_true, y_pred, zero_division=0)  # TPR
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0.0
    precision  = precision_score(y_true, y_pred, zero_division=0)
    npv        = tn / (tn + fn) if (tn + fn) > 0 else 0.0

    metrics = {
        # core
        "Accuracy":   float(accuracy_score(y_true, y_pred)),
        "Precision":  float(precision),
        "Recall":     float(sensitivity),
        "F1_Score":   float(f1_score(y_true, y_pred, zero_division=0)),
        "ROC_AUC":    float(roc_auc_score(y_true, proba_pos)) if len(np.unique(y_true)) == 2 else np.nan,
        "LOSS":       float(log_loss(y_true, proba_pos)),

        # extended (saved to CSV)
        "Balanced_Accuracy": float(balanced_accuracy_score(y_true, y_pred)),
        "Specificity":       float(specificity),
        "NPV":               float(npv),
        "PR_AUC":            float(average_precision_score(y_true, proba_pos)),
        "Brier_Score":       float(brier_score_loss(y_true, proba_pos)),
        "MCC":               float(matthews_corrcoef(y_true, y_pred)) if (tp+tn+fp+fn) > 0 else 0.0,
        "Cohen_Kappa":       float(cohen_kappa_score(y_true, y_pred)),
        "TN": int(tn), "FP": int(fp), "FN": int(fn), "TP": int(tp),
        "Threshold": float(threshold),
    }
    return metrics


def _print_fold_table(fold_idx: int, rows: list[dict]):
    """Pretty per-fold table in your requested format & order."""
    print("\n   Model                 Acc   Prec    Rec     F1    AUC")
    print("   " + "-"*54)
    order = [
        "LogisticRegression","RidgeClassifier","LDA","SVC","KNN","MLP",
        "RandomForest","GradientBoosting","AdaBoost","Bagging","ExtraTrees",
        "TabPFN_Base",
        "Ensemble_TabPFN_SVC_RF","Ensemble_TabPFN_Top5","Ensemble_AllProbModels",
        "Ensemble_Weighted_PFN+ML","Ensemble_Weighted_PFN+ML+PHE","AutoTabPFN_PHE"
    ]
    df = pd.DataFrame([r for r in rows if r["Fold"] == fold_idx])
    for name in order:
        row = df.loc[df["Model"] == name]
        if len(row) == 1:
            m = row.iloc[0]
            print(f"   {name:<22} {m['Accuracy']:6.3f} {m['Precision']:6.3f} {m['Recall']:6.3f} {m['F1_Score']:6.3f} {m['ROC_AUC']:6.3f}")


def _print_summary_mean_std(table_df_mean: pd.DataFrame, table_df_std: pd.DataFrame, title: str):
    """Print a table with mean ± std in your compact layout."""
    print("\n" + "="*60)
    print(title)
    print("="*60)
    print("Model                   Acc         Prec        Rec         F1          AUC")
    print("   " + "-"*70)
    for model in table_df_mean.index:
        m = table_df_mean.loc[model]
        s = table_df_std.loc[model]
        acc = f"{m['Accuracy']:0.3f}±{s['Accuracy']:0.3f}"
        pre = f"{m['Precision']:0.3f}±{s['Precision']:0.3f}"
        rec = f"{m['Recall']:0.3f}±{s['Recall']:0.3f}"
        f1  = f"{m['F1_Score']:0.3f}±{s['F1_Score']:0.3f}"
        auc = f"{m['ROC_AUC']:0.3f}±{s['ROC_AUC']:0.3f}"
        print(f"   {model:<21} {acc:>11}  {pre:>11}  {rec:>11}  {f1:>11}  {auc:>11}")


# ---------- main runner ----------
def run_experiment(
    data_path: str | Path,
    out_dir: str | Path,
    target_col: str,
    id_col: str,
    n_splits: int = 5,
    reps: int = 1,                     # NEW: repeated CV to boost power (e.g., 10)
    random_state: int = 42,
    device: str = "auto",              # 'auto' | 'cpu' | 'cuda' | 'mps'
    use_tabpfn_extensions: bool = True,
    enable_phe: bool = False,          # AutoTabPFN PHE (CUDA-only, optional)
    suppress_telemetry: bool = True
) -> dict:
    """
    Runs CV with TabPFN + traditional models + soft-vote ensembles.
    Saves per-fold metrics (incl. ensembles) and a rich mean±std summary.

    Returns:
        dict with paths and DataFrames: {
            'results_df': DataFrame of per-(rep,fold) metrics,
            'summary_full_path': path to summary CSV,
            'results_path_full': path to results CSV,
            'core_path': path to core metrics CSV,
        }
    """
    # optional: mute telemetry (helps avoid PostHog errors)
    if suppress_telemetry:
        os.environ.setdefault("POSTHOG_DISABLED", "1")
        os.environ.setdefault("AUTOGluon_TELEMETRY", "0")  # harmless if unused

    data_path = Path(data_path)
    out_dir = Path(out_dir)
    split_dir = out_dir / "splits"
    out_dir.mkdir(parents=True, exist_ok=True)
    split_dir.mkdir(parents=True, exist_ok=True)

    # versions & device
    try:
        from importlib.metadata import version as _pkg_version
        print("TabPFN version:", _pkg_version("tabpfn"))
    except Exception:
        print("TabPFN version: <unknown>")

    eff_device = _get_effective_device(device)
    print("Torch device (effective):", eff_device)

    # optional AutoTabPFN (PHE)
    AutoTabPFNClassifier = _maybe_import_autotabpfn(use_tabpfn_extensions)

    # data
    print("\n" + "="*60)
    print("DATA LOADING")
    print("="*60)
    df = pd.read_csv(data_path)
    assert target_col in df and id_col in df, f"{target_col} or {id_col} missing in {data_path}"
    assert df[id_col].is_unique, f"Duplicate IDs in {id_col}"

    X_df = df.drop(columns=[target_col, id_col])
    X = X_df.values
    y = df[target_col].values
    feature_names = X_df.columns.tolist()

    print(f"✅ Loaded {len(y)} samples, {X.shape[1]} features")
    print(f"   Device (effective): {eff_device}")
    print(f"   Class distribution: 0={sum(y==0)} ({sum(y==0)/len(y):.1%}), 1={sum(y==1)} ({sum(y==1)/len(y):.1%})")

    # base models (CPU-friendly where possible)
    traditional_models = {
        "LogisticRegression": make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000, random_state=random_state)),
        "RidgeClassifier":    make_pipeline(StandardScaler(), RidgeClassifier(random_state=random_state)),
        "LDA":                LinearDiscriminantAnalysis(),
        "SVC":                make_pipeline(StandardScaler(), SVC(probability=True, random_state=random_state)),
        "KNN":                make_pipeline(StandardScaler(), KNeighborsClassifier()),
        "MLP":                make_pipeline(StandardScaler(), MLPClassifier(max_iter=2000, random_state=random_state, early_stopping=True)),
        "RandomForest":       RandomForestClassifier(n_estimators=100, random_state=random_state, n_jobs=-1),
        "GradientBoosting":   GradientBoostingClassifier(n_estimators=100, random_state=random_state),
        "AdaBoost":           AdaBoostClassifier(n_estimators=100, random_state=random_state),
        "Bagging":            BaggingClassifier(n_estimators=100, random_state=random_state, n_jobs=-1),
        "ExtraTrees":         ExtraTreesClassifier(n_estimators=150, random_state=random_state, n_jobs=-1),
    }

    # containers
    all_fold_results: list[dict] = []

    # === Repeated CV ===
    for rep in range(reps):
        seed_rep = random_state + rep
        skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=seed_rep)

        print("\n" + "="*60)
        print(f"CROSS-VALIDATION (rep {rep+1}/{reps}, {n_splits} folds, stratified, seed={seed_rep})")
        print("="*60)

        def export_split_files(rep_idx: int, fold_idx: int, train_idx: np.ndarray, test_idx: np.ndarray):
            fold_dir = split_dir / f"rep_{rep_idx}" / f"fold_{fold_idx}"
            fold_dir.mkdir(parents=True, exist_ok=True)
            np.save(fold_dir / "train_idx.npy", train_idx)
            np.save(fold_dir / "test_idx.npy", test_idx)
            df.iloc[train_idx][[id_col, target_col]].to_csv(fold_dir / "train_ids.csv", index=False)
            df.iloc[test_idx][[id_col, target_col]].to_csv(fold_dir / "test_ids.csv", index=False)

        def log_row(rep_idx: int, fold_idx: int, model_name: str, metrics: dict):
            row = {"Rep": rep_idx, "Fold": fold_idx, "Model": model_name}
            row.update(metrics)
            all_fold_results.append(row)

        for fold_idx, (train_idx, test_idx) in enumerate(skf.split(X, y), 1):
            print(f"\n{'='*25} REP {rep+1}/{reps} • FOLD {fold_idx}/{n_splits} {'='*25}")
            export_split_files(rep+1, fold_idx, train_idx, test_idx)

            X_train, X_test = X[train_idx], X[test_idx]
            y_train, y_test = y[train_idx], y[test_idx]
            print(f"Train: {len(y_train)} (0={np.sum(y_train==0)}, 1={np.sum(y_train==1)})")
            print(f"Test : {len(y_test)}  (0={np.sum(y_test==0)}, 1={np.sum(y_test==1)})")

            # ---- Base TabPFN (constructor minimal for 2.2.1) ----
            p_tabpfn = None
            try:
                try:
                    base_tabpfn = TabPFNClassifier(device=eff_device)
                except Exception:
                    base_tabpfn = TabPFNClassifier(device="cpu")

                base_tabpfn.fit(X_train, y_train)
                p_tabpfn = _get_probabilities(base_tabpfn, X_test)
                met_tabpfn = _compute_all_metrics(y_test, p_tabpfn, threshold=0.5)
                log_row(rep+1, fold_idx, "TabPFN_Base", met_tabpfn)
            except Exception as e:
                print(f"   • ❌ TabPFN_Base failed on rep {rep+1} fold {fold_idx}: {e}")

            # ---- Traditional models ----
            fold_probas = {}
            if p_tabpfn is not None:
                fold_probas["TabPFN_Base"] = p_tabpfn
            strong_set = set()

            for name, model in traditional_models.items():
                try:
                    model.fit(X_train, y_train)
                    proba = _get_probabilities(model, X_test)
                    if not np.allclose(proba, proba[0]):
                        strong_set.add(name)
                    fold_probas[name] = proba
                    m = _compute_all_metrics(y_test, proba, threshold=0.5)
                    log_row(rep+1, fold_idx, name, m)
                except Exception as e:
                    print(f"   • ❌ {name} failed on rep {rep+1} fold {fold_idx}: {e}")

            # ---- Optional: AutoTabPFN Post-Hoc Ensemble (PHE) ----
            has_phe = False
            AutoTabPFNCls = _maybe_import_autotabpfn(use_tabpfn_extensions) if ("AutoTabPFNClassifier" not in globals()) else None
            AutoTabPFNCls = AutoTabPFNCls or (globals().get("AutoTabPFNClassifier", None))

            if enable_phe and (AutoTabPFNCls is not None) and (eff_device == "cuda"):
                try:
                    phe_path = out_dir / f"rep_{rep+1}" / f"fold_{fold_idx}" / "phe_workspace"
                    rmtree(phe_path, ignore_errors=True)
                    phe_fit_args = _sanitize_autogluon_fit_kwargs(dict())
                    auto_tabpfn = AutoTabPFNCls(
                        n_estimators=16,
                        balance_probabilities=True,
                        ignore_pretraining_limits=False,
                        n_ensemble_models=8,
                        phe_init_args=dict(path=str(phe_path)),
                        phe_fit_args=phe_fit_args,
                        random_state=seed_rep + fold_idx
                    )
                    auto_tabpfn.fit(X_train, y_train)
                    p_autopfn = auto_tabpfn.predict_proba(X_test)[:, 1]
                    fold_probas["AutoTabPFN_PHE"] = p_autopfn
                    m_phe = _compute_all_metrics(y_test, p_autopfn)
                    log_row(rep+1, fold_idx, "AutoTabPFN_PHE", m_phe)
                    has_phe = True
                except Exception as e:
                    print(f"   • ⚠️ AutoTabPFN_PHE failed on rep {rep+1} fold {fold_idx}: {e}")

            # ========= Ensembles (robust saving) =========
            def soft_vote(available_names: list[str]) -> np.ndarray:
                mats = [fold_probas[n] for n in available_names]
                return np.vstack(mats).mean(axis=0)

            def weighted_soft_vote(weights: dict[str, float]) -> np.ndarray:
                names = list(weights.keys())
                Z = np.zeros_like(next(iter(fold_probas.values())))
                wsum = 0.0
                for n in names:
                    if n not in fold_probas:
                        continue
                    Z += weights[n] * fold_probas[n]
                    wsum += weights[n]
                if wsum <= 0:
                    raise ValueError("No valid members for weighted ensemble.")
                return Z / wsum

            # Desired memberships (ideal)
            trio_ideal = ["TabPFN_Base", "SVC", "RandomForest"]
            top5_ideal = ["TabPFN_Base", "SVC", "RandomForest", "MLP", "ExtraTrees"]

            # Filter by what actually ran on this fold
            trio = [m for m in trio_ideal if m in fold_probas]
            top5 = [m for m in top5_ideal if m in fold_probas]
            all_prob_models = ["TabPFN_Base"] + sorted([m for m in fold_probas.keys() if m != "TabPFN_Base"])

            # Weighted ensemble (only include available)
            weights = {"TabPFN_Base": 1.5, "SVC": 1.0, "RandomForest": 1.0, "MLP": 1.0, "LogisticRegression": 0.7}
            if has_phe:
                weights["AutoTabPFN_PHE"] = 2.0
            weights = {k: v for k, v in weights.items() if k in fold_probas}
            if len(weights) >= 2:
                try:
                    p_w = weighted_soft_vote(weights)
                    m_w = _compute_all_metrics(y_test, p_w)
                    name_w = "Ensemble_Weighted_PFN+ML" + ("+PHE" if has_phe else "")
                    log_row(rep+1, fold_idx, name_w, m_w)
                    print(f"   • {name_w}: members={list(weights.keys())}")
                except Exception as e:
                    print(f"   • ⚠️ Weighted ensemble skipped: {e}")
            else:
                print(f"   • ⚠️ Ensemble_Weighted_PFN+ML(+PHE): not enough members on rep {rep+1} fold {fold_idx}")

            # Soft-vote ensembles (require ≥2 members)
            ensembles_plan = {
                "Ensemble_TabPFN_SVC_RF": trio,
                "Ensemble_TabPFN_Top5": top5,             # <-- TACO
                "Ensemble_AllProbModels": all_prob_models,
            }
            for ens_name, members in ensembles_plan.items():
                uniq_members = sorted(set(members))
                if len(uniq_members) >= 2:
                    try:
                        p_ens = soft_vote(uniq_members)
                        m_ens = _compute_all_metrics(y_test, p_ens)
                        log_row(rep+1, fold_idx, ens_name, m_ens)
                        print(f"   • {ens_name}: members={uniq_members}")
                    except Exception as e:
                        print(f"   • ⚠️ {ens_name} skipped: {e}")
                else:
                    print(f"   • ⚠️ {ens_name}: not enough members on rep {rep+1} fold {fold_idx} (got {uniq_members})")

            # pretty per-fold table (per rep)
            _print_fold_table(fold_idx, [r for r in all_fold_results if r["Rep"] == rep+1])

    # ========= Save Results =========
    print("\n" + "="*60)
    print("SAVING RESULTS")
    print("="*60)

    results_df = pd.DataFrame(all_fold_results)
    # tidy ordering for deterministic pairing later
    results_df = results_df.sort_values(["Model", "Rep", "Fold"]).reset_index(drop=True)

    results_path_full = out_dir / "cv_results_with_ensembles_all_metrics.csv"
    results_df.to_csv(results_path_full, index=False)
    print(f"✅ Saved full metrics to {results_path_full}")

    # core metrics CSV (ordered)
    core_cols = ["Rep","Fold","Model","Accuracy","Precision","Recall","F1_Score","ROC_AUC","LOSS"]
    core_df = results_df[core_cols].copy()
    core_path = out_dir / "cv_results_core_metrics.csv"
    core_df.to_csv(core_path, index=False)
    print(f"✅ Saved core metrics (ordered) to {core_path}")

    # ===== Summaries (mean ± std in console) =====
    def _summary(df: pd.DataFrame, models: list[str] | None = None):
        g_mean = df.groupby("Model")[["Accuracy","Precision","Recall","F1_Score","ROC_AUC"]].mean()
        g_std  = df.groupby("Model")[["Accuracy","Precision","Recall","F1_Score","ROC_AUC"]].std()
        if models is not None:
            existing = [m for m in models if m in g_mean.index]
            g_mean = g_mean.loc[existing]
            g_std  = g_std.loc[existing]
        return g_mean, g_std

    # Traditional models
    trad_order = [
        "LogisticRegression","RidgeClassifier","LDA","SVC","KNN","MLP",
        "RandomForest","GradientBoosting","AdaBoost","Bagging","ExtraTrees"
    ]
    mean_trad, std_trad = _summary(results_df, trad_order)
    _print_summary_mean_std(mean_trad, std_trad, "SUMMARY (Mean ± Std across folds) — Traditional models")

    # All models (order by mean ROC_AUC desc)
    mean_all, std_all = _summary(results_df)
    order_all = mean_all.sort_values("ROC_AUC", ascending=False).index
    mean_all = mean_all.loc[order_all]
    std_all  = std_all.loc[order_all]
    _print_summary_mean_std(mean_all, std_all, "SUMMARY (Mean ± Std across folds) — All models (incl. TabPFN/Ensembles)")

    # rich summary CSV (mean & std for many metrics)
    metric_cols = [
        "Accuracy","Balanced_Accuracy","Precision","Recall","Specificity","NPV",
        "F1_Score","ROC_AUC","PR_AUC","LOSS","Brier_Score","MCC","Cohen_Kappa"
    ]
    summary_full = results_df.groupby("Model")[metric_cols].agg(["mean", "std"])
    summary_full_flat = summary_full.copy()
    summary_full_flat.columns = ['_'.join(col).strip() for col in summary_full_flat.columns.values]
    summary_full_path = out_dir / "cv_summary_with_ensembles_all_metrics.csv"
    summary_full_flat.to_csv(summary_full_path)
    print(f"\n✅ Saved full summary statistics to {summary_full_path}")

    # manifest
    (out_dir / "run_manifest.json").write_text(json.dumps({
        "data_path": str(data_path.resolve()),
        "out_dir": str(out_dir.resolve()),
        "target_col": target_col,
        "id_col": id_col,
        "n_splits": n_splits,
        "reps": reps,
        "random_state": random_state,
        "device_requested": device,
        "device_effective": eff_device,
        "enable_phe": bool(enable_phe),
        "use_tabpfn_extensions": bool(use_tabpfn_extensions),
        "suppress_telemetry": bool(suppress_telemetry),
    }, indent=2))

    print("\n✅ Cross-validation + Ensembles complete!")
    print(f"   Results saved to: {out_dir}")

    return {
        "results_df": results_df,
        "results_path_full": results_path_full,
        "core_path": core_path,
        "summary_full_path": summary_full_path,
    }
```

## **All Metrics**

```python
import numpy as np
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
from scipy.stats import mannwhitneyu, wilcoxon
from statsmodels.stats.multitest import multipletests

# ------------------------- helpers -------------------------
def a12(x, y):
    """
    Vargha–Delaney A12 effect size:
    Probability that a randomly chosen x exceeds a randomly chosen y (ties count 0.5).
    """
    x = np.asarray(x, float)
    y = np.asarray(y, float)
    nx, ny = len(x), len(y)
    ranks = pd.Series(np.concatenate([x, y])).rank(method="average").to_numpy()
    rx = ranks[:nx].sum()
    return (rx / nx - (nx + 1) / 2.0) / ny

def compare_to_taco(taco_df, model_df, metric, pair_keys=("Rep","Fold","Seed")):
    """
    Compare TACO vs a single model on a metric.
    Uses paired Wilcoxon on matched (Rep/Fold/Seed) if possible; otherwise Mann–Whitney U.
    Returns (p_value, test_used, a12_effect, n_taco, n_model).
    """
    pair_cols = [c for c in pair_keys if c in taco_df.columns and c in model_df.columns]
    if pair_cols:
        merged = pd.merge(
            taco_df[pair_cols + [metric]],
            model_df[pair_cols + [metric]],
            on=pair_cols, how="inner", suffixes=("_taco","_model")
        )
        x = merged[f"{metric}_taco"].to_numpy()
        y = merged[f"{metric}_model"].to_numpy()
        if len(x) >= 5 and np.any(np.abs(x - y) > 1e-12):
            try:
                _, p = wilcoxon(x, y, zero_method="wilcox", alternative="two-sided")
                test_used = "Wilcoxon (paired)"
            except ValueError:
                _, p = mannwhitneyu(x, y, alternative="two-sided")
                test_used = "Mann–Whitney U"
        else:
            _, p = mannwhitneyu(x, y, alternative="two-sided")
            test_used = "Mann–Whitney U"
        eff = a12(x, y) if len(x) and len(y) else np.nan
        return p, test_used, eff, len(x), len(y)
    else:
        x = taco_df[metric].to_numpy()
        y = model_df[metric].to_numpy()
        _, p = mannwhitneyu(x, y, alternative="two-sided")
        return p, "Mann–Whitney U", a12(x, y), len(x), len(y)

def stars(p):
    return '***' if p <= 1e-3 else '**' if p <= 1e-2 else '*' if p <= 5e-2 else ''

# For metrics where LOWER is better (everything else assumed higher=better).
LOWER_IS_BETTER = {"LOSS", "Brier_Score"}

# ------------------------- main -------------------------
def create_taco_comparison_boxplots(
    taco_summary_csv_path: str,                # summary with *_mean/_std (411DCE)
    benchmark_results_csv_path: str,           # fold-level results (500MVG)
    taco_results_csv_path: str | None = None,  # OPTIONAL: fold-level 411DCE containing Ensemble_TabPFN_Top5
    output_dir: str = "./taco_final_plots",
    metrics=("Accuracy","Precision","Recall","F1_Score","ROC_AUC"),
    save_prefix="TACO_Comparison",
    show=True, save_png=True, save_pdf=True, dpi=300,
    use_fdr=True,
    sort_by=("median", "mean")   # NEW: will output both sorts by default
):
    """
    Builds one or two plots per metric (by median and/or mean).
    - Boxes represent fold-level vectors when available.
    - Ordering is by requested statistic (median/mean).
      * Descending for higher-is-better
      * Ascending for lower-is-better (LOSS, Brier_Score)
    - TACO is included in the same ordering (not pinned first).
    - Horizontal dashed line = TACO mean.
    - Paired tests vs. TACO with BH-FDR; A12 effect sizes saved to CSV; stars on plot.
    """
    Path(output_dir).mkdir(parents=True, exist_ok=True)
    taco_sum = pd.read_csv(taco_summary_csv_path)
    bench = pd.read_csv(benchmark_results_csv_path)

    # Optional: fold-level TACO to plot as a box (preferred)
    taco_results = None
    if taco_results_csv_path and Path(taco_results_csv_path).exists():
        tmp = pd.read_csv(taco_results_csv_path)
        if "Model" in tmp.columns and "Ensemble_TabPFN_Top5" in tmp["Model"].unique():
            taco_results = tmp[tmp["Model"] == "Ensemble_TabPFN_Top5"].copy()
            print("Using fold-level TACO results for boxplots.")
        else:
            print("Note: Ensemble_TabPFN_Top5 not found in taco_results_csv_path; TACO will be summary point.")

    bench_models = [
        "RandomForest","SVC","ExtraTrees","Bagging","GradientBoosting",
        "KNN","MLP","AdaBoost","LogisticRegression","RidgeClassifier","LDA","TabPFN_Base"
    ]
    bench = bench[bench["Model"].isin(bench_models)].copy()

    color_taco = "#DC143C"   # red
    color_bench = "#4682B4"  # blue

    # Normalize sort_by to a tuple
    if isinstance(sort_by, str):
        sort_by = (sort_by,)
    sort_by = tuple(s.lower() for s in sort_by)

    for metric in metrics:
        mean_col = f"{metric}_mean"
        std_col  = f"{metric}_std"
        if mean_col not in taco_sum.columns:
            print(f"[{metric}] '{mean_col}' missing in TACO summary; skipping.")
            continue

        # ---------- Gather entries (TACO + benchmarks) ----------
        entries_base = []

        # TACO (prefer vector if available)
        try:
            taco_mean = float(taco_sum.loc[taco_sum["Model"]=="Ensemble_TabPFN_Top5", mean_col].iloc[0])
        except Exception:
            print(f"[{metric}] TACO summary row not found; skipping.")
            continue
        taco_std  = float(taco_sum.loc[taco_sum["Model"]=="Ensemble_TabPFN_Top5", std_col].iloc[0]) if std_col in taco_sum.columns else np.nan
        taco_vec = None
        taco_med = taco_mean

        taco_kind = "summary"
        if taco_results is not None and metric in taco_results.columns:
            vec = taco_results[metric].dropna().to_numpy()
            if vec.size >= 1:
                taco_kind = "vector"
                taco_vec = vec
                taco_mean = float(np.nanmean(vec))
                taco_med  = float(np.nanmedian(vec))
                taco_std  = float(np.nanstd(vec, ddof=1)) if vec.size > 1 else 0.0

        entries_base.append({
            "name":"TACO", "is_taco":True, "kind":taco_kind,
            "mean":taco_mean, "median":taco_med, "std":taco_std, "data":taco_vec
        })

        # Benchmarks (vectors)
        if metric in bench.columns:
            for m in sorted(bench["Model"].unique()):
                vals = bench.loc[bench["Model"]==m, metric].dropna().to_numpy()
                if vals.size:
                    entries_base.append({
                        "name":m, "is_taco":False, "kind":"vector",
                        "mean":float(np.nanmean(vals)),
                        "median":float(np.nanmedian(vals)),
                        "std":float(np.nanstd(vals, ddof=1)) if vals.size>1 else 0.0,
                        "data":vals
                    })
        else:
            for m in sorted(bench["Model"].unique()):
                sub = bench.loc[bench["Model"]==m]
                if sub.empty or pd.isna(sub.iloc[0].get(mean_col, np.nan)):
                    continue
                mean_m = float(sub.iloc[0][mean_col])
                std_m  = float(sub.iloc[0].get(std_col, np.nan))
                entries_base.append({
                    "name":m,"is_taco":False,"kind":"summary",
                    "mean":mean_m,"median":mean_m,"std":std_m,"data":None
                })

        # We’ll render one plot per requested sort key
        for sort_key in sort_by:
            entries = [dict(e) for e in entries_base]  # copy base
            # --- ORDER ---
            desc = (metric not in LOWER_IS_BETTER)  # higher is better => descending
            if sort_key == "median":
                entries.sort(key=lambda d: d["median"], reverse=desc)
                sort_suffix = "_sortedByMedian"
                sort_title = "sorted by median"
            elif sort_key == "mean":
                entries.sort(key=lambda d: d["mean"], reverse=desc)
                sort_suffix = "_sortedByMean"
                sort_title = "sorted by mean"
            else:
                raise ValueError(f"Unsupported sort key: {sort_key}")

            labels = [e["name"] for e in entries]
            pos_map = {lab:i for i,lab in enumerate(labels)}

            # --- Plot ---
            fig, ax = plt.subplots(figsize=(14, 6))

            # Boxplots for vector entries
            vec_labels = [e["name"] for e in entries if e["kind"]=="vector"]
            vec_data   = [e["data"] for e in entries if e["kind"]=="vector"]
            vec_pos    = [pos_map[lab] for lab in vec_labels]

            if vec_data:
                bp = ax.boxplot(
                    vec_data, positions=vec_pos, widths=0.6, patch_artist=True, showmeans=False,
                    medianprops=dict(color="black", linewidth=1.5),
                    boxprops=dict(linewidth=1.0), whiskerprops=dict(linewidth=1.0),
                    capprops=dict(linewidth=1.0),
                    flierprops=dict(marker="o", markersize=4, markerfacecolor="gray",
                                    markeredgecolor="gray", alpha=0.5)
                )
                for i, patch in enumerate(bp["boxes"]):
                    lab = vec_labels[i]
                    patch.set_facecolor(color_taco if lab=="TACO" else color_bench)
                    patch.set_alpha(0.85)
                    patch.set_edgecolor("black")

            # Summary-only entries (points)
            for e in entries:
                if e["kind"] == "summary":
                    x = pos_map[e["name"]]
                    ax.plot([x], [e["mean"]], marker="o", markersize=8,
                            color=(color_taco if e["is_taco"] else color_bench))
                    if np.isfinite(e["std"]):
                        ax.errorbar(x, e["mean"], yerr=e["std"], fmt="none",
                                    linewidth=1.5, capsize=5,
                                    color=(color_taco if e["is_taco"] else color_bench))

            # Axes/ticks/grid
            ax.set_xticks(range(len(labels)))
            ax.set_xticklabels(labels, rotation=45, ha="right")
            y_label = metric.replace("_"," ")
            if metric in LOWER_IS_BETTER:
                y_label += " (lower is better)"
            ax.set_ylabel(y_label, fontsize=12)
            ax.grid(True, alpha=0.2, axis="y", linestyle="-", linewidth=0.5)
            ax.set_axisbelow(True)

            # y-limits from raw values (fallback to TACO mean)
            available_vals = []
            if vec_data:
                available_vals.append(np.concatenate([v for v in vec_data if v is not None and len(v)]))
            else:
                available_vals.append(np.array([entries[0]["mean"]]))
            raw_vals = np.concatenate(available_vals)
            y_min, y_max = float(np.nanmin(raw_vals)), float(np.nanmax(raw_vals))
            y_rng = max(1e-8, y_max - y_min)
            ax.set_ylim([y_min - 0.05*y_rng, y_max + 0.12*y_rng])

            # --- Horizontal line at TACO MEAN ---
            taco_entry = next(e for e in entries if e["name"] == "TACO")
            ax.axhline(taco_entry["mean"], linestyle="--", linewidth=1.25, alpha=0.9, color="gray")
            ax.text(
                0.995,
                (taco_entry["mean"] - ax.get_ylim()[0])/(ax.get_ylim()[1]-ax.get_ylim()[0]),
                f"  TACO mean = {taco_entry['mean']:.3f}",
                va="center", ha="right", transform=ax.transAxes, fontsize=10, alpha=0.9
            )

            # --- Significance vs TACO (paired where possible) ---
            if taco_entry["kind"] == "vector" and len(taco_entry["data"]) >= 5 and metric in bench.columns:
                needed_cols = [c for c in ["Rep","Fold","Seed", metric] if c in taco_results.columns]
                if metric not in needed_cols:
                    needed_cols.append(metric)
                taco_df_for_tests = taco_results[needed_cols].copy()

                pvals = []; mdl = []; tests=[]; a12s=[]; n_t=[]; n_m=[]
                tops = {}

                for e in entries:
                    if e["is_taco"] or e["kind"] != "vector":
                        continue
                    model_df = bench[bench["Model"]==e["name"]]
                    p, test_used, eff, nt, nm = compare_to_taco(
                        taco_df=taco_df_for_tests.rename(columns={metric:metric}),
                        model_df=model_df, metric=metric
                    )
                    pvals.append(p); mdl.append(e["name"]); tests.append(test_used)
                    a12s.append(eff); n_t.append(nt); n_m.append(nm)
                    tops[e["name"]] = np.max(e["data"])

                if pvals:
                    if use_fdr:
                        reject, p_adj, _, _ = multipletests(pvals, method="fdr_bh")
                    else:
                        reject = [p <= 0.05 for p in pvals]
                        p_adj = np.asarray(pvals, float)

                    sig_df = pd.DataFrame({
                        "Model": mdl,
                        "metric": [metric]*len(mdl),
                        "sorted_by": [sort_key]*len(mdl),
                        "test": tests,
                        "p_raw": pvals,
                        "p_adj_BH": p_adj,
                        "reject_0.05": reject,
                        "A12_TACO_vs_Model": a12s,
                        "n_taco": n_t,
                        "n_model": n_m
                    })
                    sig_path = Path(output_dir)/f"{save_prefix}_{metric}{sort_suffix}_significance_vs_TACO.csv"
                    sig_df.to_csv(sig_path, index=False)
                    print(f"  Significance saved: {sig_path}")

                    # Stars above the boxes
                    for m, pval in zip(mdl, p_adj):
                        if m in tops:
                            i = pos_map[m]
                            ax.text(i, tops[m] + 0.04*y_rng, stars(pval),
                                    ha="center", va="bottom", fontsize=12, fontweight="bold")
            else:
                if taco_entry["kind"] != "vector":
                    print(f"[{metric} | {sort_title}] TACO is summary-only; significance skipped.")
                elif len(taco_entry["data"]) < 5:
                    print(f"[{metric} | {sort_title}] TACO vector has <5 points; significance skipped.")

            # Save
            plt.tight_layout()
            base = f"{save_prefix}_{metric}{sort_suffix}"
            if save_png: plt.savefig(Path(output_dir)/f"{base}.png", dpi=dpi, bbox_inches="tight")
            if save_pdf: plt.savefig(Path(output_dir)/f"{base}.pdf", dpi=dpi, bbox_inches="tight")
            if show: plt.show()
            else: plt.close()

# ------------------------- runner -------------------------
def prepare_data_files():
    taco_summary = "411DCE_Genes/cv_summary_with_ensembles_all_metrics.csv"
    bench_results = "500MVG_Genes/cv_results_with_ensembles_all_metrics.csv"
    # Use ONLY if the 411DCE results file truly contains the fold-level rows for Ensemble_TabPFN_Top5
    taco_results = "411DCE_Genes/cv_results_with_ensembles_all_metrics.csv"
    return taco_summary, bench_results, taco_results

if __name__ == "__main__":
    taco_summary_file, bench_file, taco_results_file = prepare_data_files()
    create_taco_comparison_boxplots(
        taco_summary_csv_path=taco_summary_file,
        benchmark_results_csv_path=bench_file,
        taco_results_csv_path=taco_results_file,  # set to None if unavailable
        output_dir="./taco_final_plots",
        metrics=["Accuracy","Precision","Recall","F1_Score","ROC_AUC"],
        save_prefix="TACO_Comparison",
        show=True, save_png=True, save_pdf=True, dpi=300, use_fdr=True,
        sort_by=("median", "mean")   # <- both plots per metric
    )
```

## **Some Metrics**

```python
import numpy as np
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt

def _collect_fold_vectors_for_panel(taco_results_csv_path, benchmark_results_csv_path, metric):
    """Return dict: name -> np.array of fold values (only models with vectors)."""
    bench = pd.read_csv(benchmark_results_csv_path)
    need_cols = {"Model", metric}
    if not need_cols.issubset(bench.columns):
        raise ValueError(f"{benchmark_results_csv_path} must have columns {need_cols}")

    # keep only benchmark models you plot elsewhere (no ensembles, except TabPFN_Base allowed)
    bench_keep = [
        "RandomForest","SVC","ExtraTrees","Bagging","GradientBoosting",
        "KNN","MLP","AdaBoost","LogisticRegression","RidgeClassifier","LDA","TabPFN_Base"
    ]
    bench = bench[bench["Model"].isin(bench_keep)].copy()

    # build vectors for benchmarks
    vecs = {}
    for m, g in bench.groupby("Model"):
        v = g[metric].dropna().to_numpy()
        if v.size:
            vecs[m] = v

    # TACO (Ensemble_TabPFN_Top5) must come from fold-level CSV (not the summary)
    if taco_results_csv_path and Path(taco_results_csv_path).exists():
        tr = pd.read_csv(taco_results_csv_path)
        if "Model" in tr.columns and (tr["Model"] == "Ensemble_TabPFN_Top5").any() and metric in tr.columns:
            v = tr.loc[tr["Model"]=="Ensemble_TabPFN_Top5", metric].dropna().to_numpy()
            if v.size:
                vecs["TACO"] = v  # red box
        else:
            raise ValueError("TACO fold-level vectors not found (need Ensemble_TabPFN_Top5 in the results CSV).")
    else:
        raise ValueError("taco_results_csv_path missing or not found.")

    return vecs

def _rank_top5_by_mean(vecs):
    """Return list of (name, mean) sorted by mean desc, top-5."""
    stats = [(name, float(np.nanmean(v))) for name, v in vecs.items()]
    # do not let numpy compare nans oddly; all means should be finite here
    stats.sort(key=lambda t: t[1], reverse=True)
    return stats[:5]

def _small_mean_box(ax, text):
    # Top-right inside the axes
    ax.text(
        0.98, 0.96, text,
        transform=ax.transAxes, ha="right", va="top",
        fontsize=10, bbox=dict(boxstyle="round,pad=0.3", facecolor="white", edgecolor="gray", alpha=0.8)
    )

def panel_top5_by_mean(
    taco_results_csv_path,
    benchmark_results_csv_path,
    metrics=("Precision","F1_Score","ROC_AUC"),
    out_path_png="panel_top5_by_mean.png",
    out_path_pdf="panel_top5_by_mean.pdf",
    dpi=300,
):
    # colors
    color_taco = "#DC143C"   # red
    color_bench = "#4682B4"  # blue

    fig, axes = plt.subplots(1, len(metrics), figsize=(18, 4), constrained_layout=True)

    for ax, metric in zip(axes, metrics):
        # 1) collect the exact fold vectors that will be plotted
        vecs_all = _collect_fold_vectors_for_panel(taco_results_csv_path, benchmark_results_csv_path, metric)

        # 2) choose top-5 by mean (from those vectors)
        top5 = _rank_top5_by_mean(vecs_all)

        # 3) order labels by mean desc (this controls plotting order)
        labels = [name for name, _ in top5]
        means  = {name: m for name, m in top5}

        # 4) assemble vectors in that order
        data = [vecs_all[name] for name in labels]

        # 5) plot
        positions = np.arange(len(labels))
        bp = ax.boxplot(
            data, positions=positions, widths=0.6, patch_artist=True, showmeans=False,
            medianprops=dict(color="black", linewidth=1.5),
            boxprops=dict(linewidth=1.0), whiskerprops=dict(linewidth=1.0),
            capprops=dict(linewidth=1.0),
            flierprops=dict(marker="o", markersize=3.5, markerfacecolor="gray",
                            markeredgecolor="gray", alpha=0.5)
        )
        # color per box
        for i, patch in enumerate(bp["boxes"]):
            lab = labels[i]
            patch.set_facecolor(color_taco if lab == "TACO" else color_bench)
            patch.set_alpha(0.85); patch.set_edgecolor("black")

        # x ticks
        ax.set_xticks(positions)
        ax.set_xticklabels(labels, rotation=35, ha="right")
        ax.set_ylabel(metric.replace("_", " "))
        ax.grid(True, axis="y", alpha=0.2, linewidth=0.6)
        ax.set_axisbelow(True)

        # horizontal line at TACO mean (if TACO is present)
        if "TACO" in vecs_all:
            taco_mean = float(np.nanmean(vecs_all["TACO"]))
            ax.axhline(taco_mean, linestyle="--", linewidth=1.1, color="gray", alpha=0.9)
            _small_mean_box(ax, f"TACO mean = {taco_mean:.3f}")

        # light margins
        all_vals = np.concatenate(data) if len(data) else np.array([0.0])
        y_min, y_max = float(np.nanmin(all_vals)), float(np.nanmax(all_vals))
        rng = max(1e-8, y_max - y_min)
        ax.set_ylim(y_min - 0.05*rng, y_max + 0.12*rng)

        # make sure order is by mean desc
        ax.set_title(metric.replace("_", " "))

    # Save
    if out_path_png: fig.savefig(out_path_png, dpi=dpi, bbox_inches="tight")
    if out_path_pdf: fig.savefig(out_path_pdf, dpi=dpi, bbox_inches="tight")
    plt.show()
```

```python
panel_top5_by_mean(
    taco_results_csv_path="411DCE_Genes/cv_results_with_ensembles_all_metrics.csv",
    benchmark_results_csv_path="500MVG_Genes/cv_results_with_ensembles_all_metrics.csv",
    metrics=("Precision","F1_Score","ROC_AUC"),
    out_path_png="./top5_panel_by_mean.png",
    out_path_pdf="./top5_panel_by_mean.pdf",
)
```
